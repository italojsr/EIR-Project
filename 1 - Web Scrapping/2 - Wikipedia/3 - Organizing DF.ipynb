### On this notebook we are going to organize the dataframe in a MultIndex dataframe

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('Players Wikipedia.csv',index_col = 0)
```

```python
df.head()
```

```python
worldcups =  ['1930 Uruguay','1934 Italy', '1938 France', '1950 Brazil', '1954 Switzerland','1958 Sweden','1962 Chile','1970 Mexico','1982 Spain']
```

We going to slice the imported dataframe in tiny peaces for each team of each worldcup as below

```python
filtering = df['Worldcup'] == '1930 Uruguay'
filtering2 = df['Team'] == 'Bolivia'
df5 = df[filtering][filtering2]
```

```python
df5
```

The function below will do it and some more adjusts:

```python
def organize_team(df,team_name,worldcup_name):
    #df - DataFrame of shape (n,6), n - name of players
    filtering = df['Worldcup'] == worldcup_name
    filtering2 = df['Team'] == team_name
    df2 = df[filtering][filtering2]
    df2 = df2.iloc[:,2:].T
    df2 = df2.reset_index().reset_index()
    df2.columns = range(df2.shape[1])
    df2 = df2.reset_index()
    df2.columns = range(df2.shape[1])
    df2.iloc[:,0] = worldcup_name
    df2.iloc[:,1] = team_name
    return df2
    
```

We going to use this format above for each team and concat the obtained dataframes vertically in a bigger dataframe

```python
dfx = pd.DataFrame()
for worldcup in worldcups:
    teams = df[df['Worldcup'] == worldcup]['Team'].unique() #Teams from the iterated worldcup
    for team in teams:
        df2 = organize_team(df,team,worldcup)
        #df2.index = range(df2.shape[1])
        dfx = pd.concat((dfx,df2))
```

```python
dfx.head(8)
```

```python
dfx.index = range(dfx.shape[0])
dfx = dfx.fillna(0)
dfx.index = pd.MultiIndex.from_frame(dfx.iloc[:,:2])
dfx.head()
```

```python
dfx = dfx.iloc[:,2:]
dfy = dfx.T
dfy.head()
```

<b>Checking that we still with 100 players at the end:</b>

```python
n_players = 0
for worldcup in worldcups:
    teams = df[df['Worldcup'] == worldcup]['Team'].unique()
    for team in teams:
        n_players += (dfy[worldcup][team].iloc[1:,0] !=0).value_counts()[True]
```
