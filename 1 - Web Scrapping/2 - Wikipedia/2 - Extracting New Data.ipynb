### On this notebook we are going to do a new webscrapping using Wikipedia as source

```python
import pandas as pd
from bs4 import BeautifulSoup
from selenium import webdriver
import time 
import datetime
import numpy as np
from selenium.webdriver.common.by import By
```

```python
df = pd.read_csv('Missings.csv',header = [0,1])
df = df.iloc[:,1:]
df.head()
```

```python
worldcups =  ['1930 Uruguay','1934 Italy', '1938 France', '1950 Brazil', '1954 Switzerland','1958 Sweden','1962 Chile','1970 Mexico','1982 Spain']
```

Placing all players with missing data in one unique list:

```python
players_info = []
for worldcup in worldcups:
    for team in df[worldcup]:
        for player in df[worldcup][team]:
            if(player!='0'):
                players_info.append([player,team,worldcup])
        
    
```

Webscrapping:

```python
dic = {'janeiro':'01','fevereiro':'02','março':'03','abril':'04','maio':'05','junho':'06','julho':'07','agosto':'08','setembro':'09','outubro':'10','novembro':'11','dezembro':'12'}

name_and_date = []
still_missing = 0
for player_info in players_info:
    player = player_info[0]
    team = player_info[1]
    worldcup = player_info[2]
    url = 'https://www.google.com/search?q={}+jogador+de+futebol+{}'
    driver = webdriver.Chrome()
    player_url = url.format(player,team)
    driver.get(player_url)
    try:
        name_and_date.append([worldcup,team,player,driver.find_element(By.CLASS_NAME,"rVusze").text])
    except:
        print("Can't webscrapping {} from {}".format(player,team))
        still_missing += 1
        continue
    
```

```python
df = pd.read_csv('Backup.csv',index_col = 0)
dates = list(df.iloc[:,3])
dates[:10]

for i,info in enumerate(dates):
    dates[i] = dates[i].split(',')[0].split(':')[1]
```

```python
dates[:10]
```

```python
df.iloc[:,3] = dates
df.head()
```

There still some players who have the country of birth in the place of the date of birth or just the year of birth. Filtering them:

```python
df.iloc[:,3]
```

```python
undesired_indexes = []
for i,info in enumerate(dates):
    if(info.split()[0].isalpha() #If it is a country, continue
       or len(info.split()[0]) == 4): #If it is a incomplete date
        undesired_indexes.append(i)
    else:
        continue
```

```python
df = df.drop(undesired_indexes)
```

```python
still_missing2 = 188 - df.shape[0]
still_missing2
```

```python
df.index = range(df.shape[0])
```

So we can conclude that actually 88 players still have missing data, instead of the 61 it was thought

Now we need to put the date of birth in the format DD/MM/YYYY:

```python
dates2 = df.iloc[:,3]
dates3 = [x.split() for x in dates2]

day_list = []
month_list = []
year_list = []
for info in dates3:
    Day = info[0]
    Month = dic[info[2]]
    Year = info[4]
    day_list.append(Day)
    month_list.append(Month)
    year_list.append(Year)

dates2[:10]

    
```

```python
df.columns = ['Worldcup','Team','Player','Day']
df['Month'] = month_list
df['Year'] = year_list
```
