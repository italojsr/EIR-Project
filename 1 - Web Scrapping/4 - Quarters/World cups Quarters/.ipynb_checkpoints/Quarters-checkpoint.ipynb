# Goals

To create dataframes for each worldcup with the count of players borned at each trimester of the year

```python
import pandas as pd
import numpy as np     
import sys
sys.path.insert(1,'C://Users//italo//OneDrive//Documentos//GitHub//TCC//Classes')
from classes import Worldcup,Team_Players,WorldcupPlayers
```

```python
def create_quarter_df(worldcup_name,multi_columns = False, as_database = False):
    worldcup = WorldcupPlayers(worldcup_name)
    team_data = []
    teams_name = worldcup.get_teams() # Name of all teams that have participated in the selected world cup
    for team in teams_name:
        players = worldcup.get_players_from_team(team) # Instantiate Team object
        team_data.append(players.count_players_by_quarter())
        team_data.append(players.percent_players_by_quarter())
    if(multi_columns):
        quarters_df = pd.DataFrame(team_data).T
        column_names = []
        for i,column in enumerate(quarters_df.columns):
            condition = (i%2 == 0)
            if(condition == True):
                column_names.append((teams_name[int(i/2)],'Frequence'))
            else:
                column_names.append((teams_name[int(i/2)],'Percentage(%)'))
        quarters_df.columns = pd.MultiIndex.from_tuples(column_names)
        quarters_df.index = ['Total','First Trimester','Second Trimester','Third Trimester','Forth Trimester']
    if(as_database):
        quarters_df = pd.DataFrame(team_data)
        quarters_df.index = ['Frequence','Percentage(%)']*(int(quarters_df.shape[0]/2))
        quarters_df.reset_index(names = ['Players'],inplace = True)
        quarters_df.index = np.repeat(teams_name,2)
        quarters_df.columns = ['Players','Total','First Quarter','Second Quarter','Third Quarter','Forth Quarter']
        quarters_df.reset_index(names = ['Team'],inplace = True)
    return quarters_df
```

```python
worldcups = ['1930 Uruguay', '1934 Italy', '1938 France', '1950 Brazil', '1954 Switzerland', '1958 Sweden', '1962 Chile',
 '1966 England', '1970 Mexico', '1974 Germany', '1978 Argentina', '1982 Spain', '1986 Mexico', '1990 Italy', '1994 USA',
 '1998 France', '2002 Japan', '2006 Germany', '2010 South Africa', '2014 Brazil', '2018 Russia', '2022 Qatar']
```

```python
for worldcup in worldcups:
    df = create_quarter_df(worldcup,False,True)
    df.to_csv('{} quarters (DataBase).csv'.format(worldcup),index = False)
    df.to_excel('{} quarters (DataBase).xlsx'.format(worldcup),index = False)
    
```

```python
for worldcup in worldcups:
    df = create_quarter_df(worldcup,True,False)
    df.to_csv('{} quarters (MultiColumns).csv'.format(worldcup))
    df.to_excel('{} quarters (MultiColumns).xlsx'.format(worldcup))
    
```
