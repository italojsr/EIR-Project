```python
import pandas as pd
from bs4 import BeautifulSoup
from selenium import webdriver
import time 
import datetime
import numpy as np
from selenium.webdriver.common.by import By

```

```python
dados = pd.read_csv('Times.csv')
dados.head()
```

### Raspar a data de nascimento de um jogador

Escolhendo o jogador do qual se buscará o nascimento

```python
jogador = dados.iloc[0,0]
time = dados.columns[0]
```

Abre uma guia no Chrome e passa pro webdrive o url

```python
url = 'https://www.google.com/search?q={}+jogador+de+futebol+{}'
driver = webdriver.Chrome()
jogador_url = url.format(jogador,time)
driver.get(jogador_url)
```

```python
dic = {'janeiro':'01','fevereiro':'02','março':'03','abril':'04','maio':'05','junho':'06','julho':'07','agosto':'08','setembro':'09','outubro':'10','novembro':'11','dezembro':'12'}
```

Função que pega um texto que contém a data e devolve a data formatada no formato DD/MM/AA

```python
def pega_data(descricao):
    sdescricao = descricao.split()
    dia,mes,ano = sdescricao[1:6:2]
    if(int(ano)<1950):
        print('Algo que está errado não está certo')
        return '0'
    else:
        mes = mes.replace(mes,dic[mes])
        data_de_nasc = ('{}/{}/{}').format(dia,mes,ano)
        return data_de_nasc
```

Pega o texto que contém a data de nascimento do jogador da página de pesquisa

```python
descricao = driver.find_element(By.CLASS_NAME,"Z1hOCe").text
data_de_nascimento = pega_data(descricao)
print('A data de nascimento de {} do time {} é {}'.format(jogador,time,data_de_nascimento))


```

### Generalizando

```python
jogador = dados.iloc[0,0]
time = dados.columns[0]
```

```python
df_data_de_nasc = pd.DataFrame([])

for i,nome_do_time in enumerate(dados.columns):

    data_de_nascimento = np.zeros(dados.shape[0],dtype = 'U64') #Lista que vai guardar a data de nascimento para jogadores de um único time
    for j,jogador_do_time in enumerate(dados.iloc[:,i]):
        try:
            jogador = jogador_do_time
            time = nome_do_time
            url = 'https://www.google.com/search?q={}+jogador+{}'
            driver = webdriver.Chrome()
            jogador_url = url.format(jogador,time)
            driver.get(jogador_url)
            try:
                descricao = driver.find_element(By.CLASS_NAME,"Z1hOCe").text
                data_de_nascimento[j] = pega_data(descricao)
            except:
                print('Não foi possível pegar a descricao do jogador {} do time do {}'.format(jogador,time)) 
                data_de_nascimento[j] = '0'
            df_data_de_nasc[i] = data_de_nascimento
                
            
        except:
            print('Não foi possível adicionar a lista do time {}'.format(time))

        
```
