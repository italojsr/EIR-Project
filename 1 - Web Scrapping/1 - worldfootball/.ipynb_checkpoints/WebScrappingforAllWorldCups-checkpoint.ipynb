### In this notebook we going to extract information of players from all teams of since the 1930 Uruguay worldcup

```python
import pandas as pd
from bs4 import BeautifulSoup
from selenium import webdriver
import time
import datetime
import numpy as np
from selenium.webdriver.chrome.options import Options
```

<p>General Assumptions:</p>
<p>All teams must have 26 players and missing players will be placed with 0's in the dataframe </p>

#### Given a worldcup extract the name of teams and the link for each team website

```python
def team_names_and_links(worldcup):
    # Returns two numpy arrays
    #teams_en - team names in english
    #teams_ge - part of the links used to access each team website page 
    teams_ge = []
    teams_en = []
    worldcup_url = 'https://www.worldfootball.net/players/{}/'
    condition = True
    while(condition == True):
        try:
            
            
            driver = webdriver.Chrome()
            worldcup_url = worldcup_url.format(worldcup)
            driver.get(worldcup_url)
            time.sleep(10)
            soup = BeautifulSoup(driver.page_source,'html.parser')
            table = soup.find('table',{'class':'standard_tabelle'})
            trs = table.find_all('tr')
            driver.close()
        except:
            print("Soup error on team_names_and_links from {}".format(worldcup))
            driver.close()
            continue
        break
    for tr in trs:
        team_en = tr.text.split('\n')
        team_en = [x for x in team_en if (x != '') and (x.isnumeric() is False) ]
        team_en = team_en[0]
        team_ge = tr.a['href'].split('/')[2]
        teams_en.append(team_en)
        teams_ge.append(team_ge)
    teams_en = np.array(teams_en)
    teams_ge = np.array(teams_ge)
    return(teams_en,teams_ge)
```

#### Getting the information from players

```python
field_position = ['Goalkeeper','Defender','Midfielder','Forward']
```

```python
def check_date_birth_exist(info):
    last_info_player = info[-1]
    if (sum(c in '/' for c in last_info_player) == 2) and (sum(c.isdigit() for c in last_info_player) ==8):
        return True
    else:
        return False
```

```python
#def add_player(info,players,positions,player_position):
    #Organize extracted information from a player in a list
#    try:
#        player_name = info[0]
#        day,month,year = info[-1].split('/')
#        player_info = [player_name,player_position,day,month,year]
#    except:
#        print("Can't webscrapping player {}".format(info))
#   return player_info

def add_player(info,players,positions,player_position):
    #Organize extracted information from a player in a list
    try:
        player_name = info[0]
        if check_date_birth_exist(info):
            day,month,year = info[-1].split('/')
            player_info = [player_name,player_position,day,month,year]
        else:
            player_info = [player_name,player_position,'00','00','0000']
    except:
        print("Can't webscrapping player {} ".format(info[0]))
    return player_info
```

```python
def new_index(df,team):
    #Set a new index with the name of the team repeated through the given DataFrame to be used on the MultIndexed DataFrame
    new_index = []
    df_rows = df.shape[0]
    team_en = teams_en[teams_ge == team ][0]
    for i in range(df_rows):
        new_index.append(team_en)
    return new_index
```

```python
def organize_df(df,team):
    #Organize the DataFrame to be placed with other teams in a MultIndex DataFrame
    df2 = df.T.reset_index()
    df2.columns = range(df2.shape[1])
    df2.index = new_index(df2,team)
    return df2
    
```

```python
def team_webscrapping(team,worldcup):
    #Returns a dataframe which have the information from all players of a determined team
    
    condition = True
    while(condition == True):
        try:
            team_url = 'https://www.worldfootball.net/teams/{}/{}/2/'
            team_url = team_url.format(team,worldcup)
            print(team_url)

            driver = webdriver.Chrome()
            driver.get(team_url)
            time.sleep(10)
            soup = BeautifulSoup(driver.page_source,'html.parser')
            table = soup.find('table',{'class':'standard_tabelle'})
            trs = table.find_all('tr')
            driver.close()
        except:
            print("Soup don't worked with {}".format(team))
            driver.close()
            continue
        break
    players = []
    positions = []
    i = -1

    for tr in trs:
        info = tr.text.split('\n')
        info = [x for x in info if (x != '') and (x.isnumeric() is False) ] #Take out the spaces and number of each player of the list
        position = (len(info)==1) #Positions come with just one information in the list (e.g GoalKeeper)
        player = (len(info)>1)
        try:
            if(position):
                position_info = info[0]
                positions.append(position_info)
                i += 1
            elif(player):
                player_position = positions[i]
                if(player_position not in field_position):
                    continue #Not a player, but some coach position
                else:
                    players.append(add_player(info,players,positions,player_position))
        except:
            print('Not a position neither an exception')
        
        
    df = pd.DataFrame(players,columns = ['Name','Position','Day','Month','Year'])
    df = organize_df(df,team)
    return df
```

```python
def concat_teams(df,df_teams,team):
    # At each iteration will add a team to the initial DataFrame
    if(df_teams.shape[1] == df.shape[1]): #Check if the number of players on the team is the same of the dataframe
        df_teams = pd.concat([df_teams,df],axis = 0)
        #print('Same dimension')
        return df_teams
    elif(df_teams.shape[1] - df.shape[1] > 0): #If there is missing players...
        n_missing_players = df_teams.shape[1] - df.shape[1]
        n_features = df.shape[0]
        for i in range(n_missing_players):
            missing_player = pd.DataFrame(np.zeros(n_features,dtype = int),index = new_index(df,team) ) #Add players with 0's
            df = pd.concat([df,missing_player],axis = 1)
            df.columns = range(df.shape[1])
        df_teams = pd.concat([df_teams,df],axis = 0)
        print('The {} have less player than usual'.format(team))
        return df_teams
    else:
        name = df.index[0]
        print('The number of players of the {} team is different from the number in the dataframe').format(name)
        return
        
      
```

```python
def organize_multindex_df(df):
    df = df.reset_index()
    df.columns = range(df.shape[1])
    df.index = pd.MultiIndex.from_frame(df.iloc[:,:2])
    df = df.T
    df = df.iloc[2:,:]
    return df
```

```python
def create_worldcup_df(teams_ge,teams_en,worldcup,worldcup_name):
    #create a csv and excel file for each worldcup with the webscrapping data
    team_size = 26
    worldcup_df = pd.DataFrame([],columns = range(team_size + 1))
    for j,team in enumerate(teams_ge):
        try:
            df = team_webscrapping(team,worldcup)
        except:
            print('Error webscrapping {} '.format(teams_en[j]))
            break
        try:
            worldcup_df = concat_teams(df,worldcup_df,team)
        except:
            print("Couldn't add the {} ".format(teams_en[j]))
            break
    worldcup_df = organize_multindex_df(worldcup_df)
    
    if '/' in worldcup_name: #Exception for Japan/SouthKorea Worldcup
        worldcup_name = worldcup_name.split('/')[0]
    
    worldcup_df.to_csv('{}.csv'.format(worldcup_name))
    worldcup_df.to_excel('{}.xlsx'.format(worldcup_name))
    return worldcup_df
    
    
```

#### Extraction of the links for access each worldcup

```python
url = 'https://www.worldfootball.net/players/wm-2022-in-katar/'

    
driver = webdriver.Chrome()
driver.get(url)
time.sleep(10)
soup = BeautifulSoup(driver.page_source,'html.parser')
driver.close()
```

```python
worldcups_en = []
worldcups_ge = []
dropdown_div = soup.find('div',{'class':'box2'})
select = dropdown_div.find('select',{'name':'saison'})
options = select.find_all('option')
for option in options:
    worldcup_en = option.text
    worldcups_en.append(worldcup_en)
    worldcup_ge = option['value'].split('/')[2]
    worldcups_ge.append(worldcup_ge)
```

#### Worldcups WebScrapping

```python
worldcups_ge[13:14]
```

```python
for i,worldcup in enumerate(worldcups_ge):        
    worldcup_name = worldcups_en[i]
    try:
        teams_en,teams_ge = team_names_and_links(worldcup)
    except:
        print("Could not capture the name of teams from {}".format(worldcup_name))
    try:
        new_worldcup_df = create_worldcup_df(teams_ge,teams_en,worldcup,worldcup_name)
    except:
        continue

    
```
