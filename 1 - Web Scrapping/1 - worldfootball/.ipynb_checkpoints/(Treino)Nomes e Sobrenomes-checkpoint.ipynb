### Funções Para Pegar Primeiro e segundo Nome dos Jogadores de um Time

Conseguindo todos os nomes e sobrenomes dos jogadores

```python
def encontra_primeiros_nomes():
    name_list = []
    for first_name_card in soup.find_all('div',{'class' : ['player-badge-card_playerFirstname__1HlN6']}):
        try:
            name = first_name_card.find('span',{'class' : ['d-none d-md-block']}).text
            name_list.append(name)
        except:
            name_list.append('0')
    return name_list


```

```python
def encontra_sobrenomes():
    surname_list = []
    for surname_card in soup.find_all('div',{'class' : ['player-badge-card_playerSurname__3_5gf ff-mt-4']}):
        try:
            surname = surname_card.find('span',{'class' : ['d-none d-md-block']}).text
            surname_list.append(surname)
        except:
            surname_list.append(' ')
    return surname_list
        


```

```python
def junta_nome_sobrenome(name_list,surname_list):
    players = ["{} {}".format(name.lower().capitalize(),surname.lower().capitalize()) for name,surname in zip(name_list,surname_list)]
    players
    return players
```

### Encontrando todos os times da liga

```python
team_names_url = 'https://www.fifa.com/fifaplus/en/tournaments/mens/worldcup/qatar2022/teams'
driver = webdriver.Chrome()
driver.get(team_names_url)
time.sleep(10)
soup2 = BeautifulSoup(driver.page_source,'html.parser')
```

```python
team_list = []

grids = soup2.find('div',{'class':'flag-grid_flagGrid__2y0Q-'})
links = grids.find_all('a')

for link in links:
    team_name = link.find('div',{'class':['flag-with-info_flagCountry__Yw8QR']}).text
    team_list.append(team_name.lower().replace(" ","-"))  
    
team_list 
```

<p>Função abaixo deve ajustar acrescentar missing values quando o número de jogadores em um time for diferente do dataset geral</p>
<p>Porém no ínicio do looping, o dataset já tem tamanho diferente do primeiro time, dimensão zero.</p>

```python
def tratamento_excecao(i,df,players_name_list):
    tam_players = len(players_name_list) #Número de jogadores do time
    
    if (df.shape[0] == tam_players): #Checa se o número de jogadores no time é o mesmo do dataframe com os outros times
        df[i] = players_name_list
        return
    elif(df.shape[0] - tam_players > 0):
        n_jogador_faltantes = df.shape[0] - tam_players
        for i in range(n_jogador_faltantes):
            players_name_list.append('0') #Acrescenta 0 onde houver missing value no time que entrará no dataframe
        df[i] = players_name_list
        return
    else:
        print('O número de jogadores no time {} é maior que o número de jogadores no datafram').format(team_list[i])
        return
    
```
